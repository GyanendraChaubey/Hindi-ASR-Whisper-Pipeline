# Kaggle Notebook - Question 1 
 
This notebook runs Q1 in debug or full strict mode.

```python
import os 
import subprocess 
import sys 
 
REPO_DIR = '/kaggle/working/Hindi-ASR-Whisper-Pipeline' 
REPO_URL = 'https://github.com/GyanendraChaubey/Hindi-ASR-Whisper-Pipeline.git' 
 
if not os.path.exists(REPO_DIR): 
    subprocess.run(['git', 'clone', REPO_URL, REPO_DIR], check=True) 
 
os.chdir(REPO_DIR) 
subprocess.run([sys.executable, '-m', 'pip', 'install', '-r', 'requirements.txt'], check=True) 
print('Setup complete. Current dir:', os.getcwd())
```

## Run Configuration

```python
# Choose: 'debug' or 'full' 
RUN_MODE = 'debug' 
USE_CPU = False 
 
manifest_csv = 'data/FT Data - data.csv' 
device = 'cpu' if USE_CPU else 'cuda' 
 
if RUN_MODE == 'debug': 
    cmd = [ 
        sys.executable, 'scripts/run_question1.py', 
        '--manifest-csv', manifest_csv, 
        '--work-dir', 'artifacts/q1_debug', 
        '--finetuned-model-dir', 'models/whisper-small-hindi-q1-debug', 
        '--max-recordings', '4', 
        '--max-fleurs-samples', '30', 
        '--train-epochs', '1', 
        '--skip-strict-checks', 
        '--device', device, 
    ] 
else: 
    cmd = [ 
        sys.executable, 'scripts/run_question1.py', 
        '--manifest-csv', manifest_csv, 
        '--work-dir', 'artifacts/q1', 
        '--finetuned-model-dir', 'models/whisper-small-hindi-q1', 
        '--device', device, 
    ] 
 
print('Running:', ' '.join(cmd)) 
subprocess.run(cmd, check=True)
```
