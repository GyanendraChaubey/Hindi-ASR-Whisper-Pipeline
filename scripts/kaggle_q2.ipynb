# Kaggle Notebook - Question 2 
 
This notebook runs Q2 cleanup pipeline.

```python
import os 
import subprocess 
import sys 
 
REPO_DIR = '/kaggle/working/Hindi-ASR-Whisper-Pipeline' 
REPO_URL = 'https://github.com/GyanendraChaubey/Hindi-ASR-Whisper-Pipeline.git' 
 
if not os.path.exists(REPO_DIR): 
    subprocess.run(['git', 'clone', REPO_URL, REPO_DIR], check=True) 
 
os.chdir(REPO_DIR) 
subprocess.run([sys.executable, '-m', 'pip', 'install', '-r', 'requirements.txt'], check=True) 
print('Setup complete. Current dir:', os.getcwd())
```

```python
# Choose: 'debug' or 'full' 
RUN_MODE = 'full' 
USE_CPU = False 
 
manifest_csv = 'data/FT Data - data.csv' 
device = 'cpu' if USE_CPU else 'cuda' 
 
if RUN_MODE == 'debug': 
    cmd = [ 
        sys.executable, 'scripts/run_question2.py', 
        '--manifest-csv', manifest_csv, 
        '--work-dir', 'artifacts/q2_debug', 
        '--max-recordings', '4', 
        '--max-segments', '120', 
        '--device', device, 
    ] 
else: 
    cmd = [ 
        sys.executable, 'scripts/run_question2.py', 
        '--manifest-csv', manifest_csv, 
        '--work-dir', 'artifacts/q2', 
        '--device', device, 
    ] 
 
print('Running:', ' '.join(cmd)) 
subprocess.run(cmd, check=True)
```

```python
from pathlib import Path 
 
work_dir = Path('artifacts/q2' if RUN_MODE == 'full' else 'artifacts/q2_debug') 
expected = [ 
    work_dir / 'raw_asr_pairs.csv', 
    work_dir / 'q2_processed_transcripts.csv', 
    work_dir / 'number_normalization_examples.json', 
    work_dir / 'english_detection_examples.json', 
    work_dir / 'summary.json', 
    work_dir / 'question2_report.md', 
] 
 
for p in expected: 
    print('OK' if p.exists() else 'MISSING', p)
```
